### Q7. Sum of weights

```python
df = df[
    df['origin'] == 'Asia'
]
```

```python
df = df[['vehicle_weight', 'model_year']]
```

```python
df = df.head(7)
df
```

```python
import numpy as np
```

```python
X = df.values
X
```

```python
X_T = X.T
XTX = X_T.dot(X)
XTX
```

```python
XTX = np.linalg.inv(XTX)
```

```python
y = [1100, 1300, 800, 900, 1000, 1100, 1200]
```

```python
w = XTX.dot(X.T)
w = w.dot(y)
w
```

```python
w.sum()
```
